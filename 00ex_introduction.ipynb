1. The MickeyMouse problem

a) Write a program that prints the numbers from 1 to 100. But for multiples of 3 print `Mickey` instead of the corresponding number and for the multiples of 5 print `Mouse`. For numbers which are multiples of both three and five print `MickeyMouse`

b) Put the result in a tuple and substitute `Mickey` with `Donald` and `Mouse` with `Duck`

```python
def mickey_mouser():
  """
  Returns a tuple with all the numbers between 1 and 100 except:
  - "Mickey" if it is a multiple of 3
  - "Mouse" if it is a multiple of 5
  - "MickeyMouse" if it is a multiple of 15 
  """
  l = []
  for i in range(1, 101, 1):
    if i%15==0: 
      l.append("MickeyMouse")
      continue
    if i%5==0:
      l.append("Mouse")
      continue
    if i%3==0:
      l.append("Mickey")
      continue
    l.append(i)
  return tuple(l)
```

```python
mm = mickey_mouser()
print(mm)
```

```python
def donald_ducker(mm):
  """
  Transforms all instances of "Mickey" and "Mouse" in `mm` to "Donald" and "Duck"
  """
  dd = []
  for i in mm:
    if i=="MickeyMouse":
      dd.append("DonaldDuck")
      continue
    if i=="Mouse":
      dd.append("Duck")
      continue
    if i=="Mickey":
      dd.append("Donald")
      continue
    dd.append(i)
  return tuple(dd)
```

```python
dd = donald_ducker(mm)
print(dd)
```

3\. Computing the distance

Write a function that calculates and returns the euclidean distance between two points *u* and *v*, where *u* and *v* are both 2-tuples *(x,y)*. For example, if *u=(3,0)* and *v=(0,4)*, the function should return 5

```python
def euclidean_distance(u, v):
  """
  Computes the euclidean distance between two 2D points 
  """
  return pow((u[0] - v[0])**2 + (u[1] - v[1])**2, 1/2)
```

```python
u = (3, 0)
v = (0, 4)

print(euclidean_distance(u ,v))
```

4\. Counting letters

Write a program to calculate the number of times each character occurs in a given string *s*. Ignore differneces in capitalization

```python
def string_counter(s: str):
  """
  Returns a `dict` where each key is a character in the string `s` 
  and each value is how many times that character is present in `s` (ignoring capitalization)
  """
  s = s.lower()
  d = {}
  for c in s:
    if c in d.keys():
      continue
    d[c] = s.count(c)
  return d
```

```python
s = "cielO azzurro"
d = string_counter(s)
print(d)
```

5\. Isolating the unique

Write a function that determines and count the unique numbers in the list *l*

```python
def unique_isolator(l: list) -> tuple[int, list]:
  """
  Isolates all the unique values in `l` and returns how many there are
  """
  u = []
  for i in l:
    if l.count(i)==1:
      u.append(i)
  return len(u), u
```

```python
l = [1, 2, 3, 4, 1, 4]

print(unique_isolator(l))
```

6\. Combination of functions

Write two functions - one that returns the square of a number, and one that returns the cube. Now write a third function that returns the number raised to the 6th power using the two previous functions.

```python
def squarer(n):
  return n**2

def cuber(n):
  return n**3

def sixter(n):
  return cuber(squarer(n))
```

8\. Nested list comprehension

A Pythagorean triple is an integer solution to the Pythagorean theorem $a^2+b^2=c^2$. The first Pythagorean triple is (3,4,5). Find and put in a tuple all unique Pythagorean triples for the positive integers a, b and c less than 100.

```python
p = tuple([(a, b, c) for a in range(1, 101) for b in range(1, 101) for c in range(1, 101) if (a**2 + b**2 == c**2) and (a < b) and (b < c)])
print(p)
```

9\. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one

```python
def normalizer(t: tuple):
  s = 0
  for i in t:
    s += i

  return tuple([i/s for i in t])
```

```python
t = (2, 4, 6)

print(normalizer(t))
```
